```sql
-- Crear esquema Landing
USE CATALOG airline_catalog;
CREATE SCHEMA IF NOT EXISTS airline_catalog.landing
COMMENT 'Esquema para archivos';
```

```sql

-- ============================================================
-- Crear the Volume para archivos
-- Crear volume tipo MANAGED (Databricks administra el storage)
-- ============================================================

CREATE VOLUME IF NOT EXISTS airline_catalog.landing.archivos
COMMENT 'Volume para almacenar archivos CSV crudos';
-- Verificar que se creó
SHOW VOLUMES IN airline_catalog.landing;
```

```sql

-- Verificar que el archivo se subió correctamente
-- Listar archivos en el volume
LIST '/Volumes/airline_catalog/landing/archivos';
```

```sql
-- Leer directo del archivo es posible, y es útil también para analizarlo
SELECT * FROM read_files(
 '/Volumes/airline_catalog/landing/archivos/2025-01.csv',
 format => 'csv',
 header => true
 )
 limit 20
```
