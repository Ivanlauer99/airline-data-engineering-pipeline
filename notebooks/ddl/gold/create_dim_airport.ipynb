```sql
--Creacion de tabla dim_airport
DROP TABLE IF EXISTS airline_catalog.gold.dim_airport;
CREATE TABLE IF NOT EXISTS airline_catalog.gold.dim_airport
(
    airport_id BIGINT COMMENT 'id aereopuerto',
    airport_code STRING COMMENT 'Codigo de aereopuerto',
    airport_name STRING COMMENT 'Aeropuerto',
    city STRING COMMENT 'Ciudad',
    state STRING COMMENT 'Estado',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported');
```
