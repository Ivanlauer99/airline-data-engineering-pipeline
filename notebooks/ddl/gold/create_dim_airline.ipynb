```sql
--Creamos la tabla dimension de aerolíneas
DROP TABLE IF EXISTS airline_catalog.gold.dim_airline;
CREATE TABLE IF NOT EXISTS airline_catalog.gold.dim_airline
(
    airline_id BIGINT GENERATED ALWAYS AS IDENTITY,
    airline_code STRING COMMENT 'Código de aerolínea (ej: AA, DL, UA)',
    airline_name STRING COMMENT 'Nombre de la aerolínea',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported');
```
