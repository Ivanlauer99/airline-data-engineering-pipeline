```sql
-- Inserta los datos a la tabla fact_flights desde la tabla silver y las tablas dim
TRUNCATE TABLE airline_catalog.gold.fact_flights;
INSERT INTO airline_catalog.gold.fact_flights (
  flight_id, -- Identificador único del vuelo
  date_id, -- Fecha del vuelo (FK hacia dim_date)
  airline_id, -- Aerolínea operadora (FK hacia dim_airline)
  origin_airport_id, -- Aeropuerto de origen (FK hacia dim_airport)
  destination_airport_id, -- Aeropuerto de destino (FK hacia dim_airport)
  flight_number, -- Número comercial del vuelo
  tail_number, -- Identificador de la aeronave
  scheduled_departure_time, -- Hora programada de salida (HHMM)
  actual_departure_time, -- Hora real de salida (HHMM)
  departure_delay, -- Retraso en salida (minutos)
  scheduled_arrival_time, -- Hora programada de llegada (HHMM)
  actual_arrival_time, -- Hora real de llegada (HHMM)
  arrival_delay, -- Retraso en llegada (minutos)
  cancelled, -- Indica si el vuelo fue cancelado
  cancellation_code, -- Motivo de cancelación (si aplica)
  diverted, -- Indica si el vuelo fue desviado
  actual_elapsed_time, -- Duración total del vuelo en minutos
  air_time, -- Tiempo real en aire (minutos)
  distance, -- Distancia recorrida (millas)
  carrier_delay, -- Retraso causado por la aerolínea
  weather_delay, -- Retraso por clima
  nas_delay, -- Retraso por sistema aéreo/nacional
  security_delay, -- Retraso por seguridad
  late_aircraft_delay, -- Retraso por llegada tardía de aeronave anterior
  flight_status, -- Estado del vuelo (On Time, Delayed, Cancelled)
  _processing_timestamp -- Timestamp de carga del registro en el pipeline
)
SELECT
  s.flight_id, -- Identificador único del vuelo
  d.date_id, -- Fecha del vuelo
  a.airline_id, -- Aerolínea operadora
  ao.airport_id AS origin_airport_id, -- Aeropuerto de origen
  ad.airport_id AS destination_airport_id, -- Aeropuerto de destino
  s.flight_number, -- Número comercial del vuelo
  s.tail_number, -- Identificador de la aeronave
  s.scheduled_departure_time, -- Hora programada de salida
  s.actual_departure_time, -- Hora real de salida
  s.departure_delay, -- Retraso en salida
  s.scheduled_arrival_time, -- Hora programada de llegada
  s.actual_arrival_time, -- Hora real de llegada
  s.arrival_delay, -- Retraso en llegada
  s.cancelled, -- Indica si el vuelo fue cancelado
  s.cancellation_code, -- Motivo de cancelación
  s.diverted, -- Indica si el vuelo fue desviado
  s.actual_elapsed_time, -- Duración total del vuelo
  s.air_time, -- Tiempo real en aire
  s.distance, -- Distancia recorrida
  s.carrier_delay, -- Retraso causado por la aerolínea
  s.weather_delay, -- Retraso por clima
  s.nas_delay, -- Retraso por sistema aéreo/nacional
  s.security_delay, -- Retraso por seguridad
  s.late_aircraft_delay, -- Retraso por llegada tardía de aeronave anterior
  s.flight_status, -- Estado del vuelo
  current_timestamp() -- Timestamp de carga del registro
FROM airline_catalog.silver.flights_silver s -- Tabla silver de vuelos
left JOIN airline_catalog.gold.dim_date d ON s.flight_date = d.date_id -- Unión con dimensión fecha
left JOIN airline_catalog.gold.dim_airline a ON s.airline_code = a.airline_code -- Unión con dimensión aerolínea
left JOIN airline_catalog.gold.dim_airport ao ON s.origin_airport_id = ao.airport_id -- Unión con dimensión aeropuerto de origen
left JOIN airline_catalog.gold.dim_airport ad ON s.destination_airport_id = ad.airport_id -- Unión con dimensión aeropuerto de destino
```

```sql
select * from airline_catalog.gold.fact_flights
where cancellation_code is not null or carrier_delay is not null or   weather_delay is not null or nas_delay is not null or security_delay is not null or late_aircraft_delay  is not null
```
