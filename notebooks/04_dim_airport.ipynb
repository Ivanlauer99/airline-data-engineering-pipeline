```sql
--Insertar datos a dim_airport
INSERT INTO airline_catalog.gold.dim_airport (airport_code, city, state)
SELECT DISTINCT source.airport_code, source.city, source.state
FROM (

    SELECT
        origin_code AS airport_code,
        origin_city AS city,
        origin_state AS state
    FROM airline_catalog.silver.flights_silver

    UNION

    SELECT
        destination_code AS airport_code,
        destination_city AS city,
        destination_state AS state
    FROM airline_catalog.silver.flights_silver

) source

LEFT ANTI JOIN airline_catalog.gold.dim_airport target
ON source.airport_code = target.airport_code

WHERE source.airport_code IS NOT NULL;
```
