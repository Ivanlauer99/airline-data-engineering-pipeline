```sql
TRUNCATE TABLE airline_catalog.gold.dim_airport;
INSERT INTO airline_catalog.gold.dim_airport
(
    airport_id,
    airport_code,
    airport_name,
    city,
    state
)

SELECT
    a.airport_id,
    a.airport_code,
    l.description AS airport_name,
    a.city,
    a.state

FROM
(
    SELECT DISTINCT
        origin_airport_id AS airport_id,
        origin_code AS airport_code,
        origin_city AS city,
        origin_state AS state
    FROM airline_catalog.silver.flights_silver

    UNION

    SELECT DISTINCT
        destination_airport_id AS airport_id,
        destination_code AS airport_code,
        destination_city AS city,
        destination_state AS state
    FROM airline_catalog.silver.flights_silver
) a

LEFT JOIN airline_catalog.lookups.airport_lookup l
    ON a.airport_code = l.code;
```
