```python
%sql
CREATE VIEW IF NOT EXISTS airline_catalog.semantic.vw_flights AS
SELECT 
  dt.year as year, -- Año del vuelo
  dt.month as month, -- Mes del vuelo
  dt.day as day,-- Día del vuelo
  ff.flight_id, -- Identificador único del vuelo
  da.airline_name, -- Nombre de la aerolínea
  ao.airport_name AS origin_airport_name, -- Nombre del aeropuerto de origen
  ao.city AS origin_city, -- Ciudad del aeropuerto de origen
  ao.state AS origin_state, -- Estado del aeropuerto de origen
  ad.airport_name AS destination_airport_name, -- Nombre del aeropuerto de destino
  ad.city AS destination_city, -- Ciudad del aeropuerto de destino
  ad.state AS destination_state, -- Estado del aeropuerto de destino
  ff.distance, -- Distancia del vuelo
  ff.actual_elapsed_time, -- Duración total del vuelo
  ff.cancelled, -- Indica si el vuelo fue cancelado
  ff.departure_delay, -- Retraso en salida
  ff.arrival_delay, -- Retraso en llegada
  ff.flight_status -- Estado del vuelo
  
FROM airline_catalog.gold.fact_flights ff -- Tabla de hechos de vuelos
INNER JOIN airline_catalog.gold.dim_airline da -- Dimensión de aerolínea
  ON ff.airline_id = da.airline_id
INNER JOIN airline_catalog.gold.dim_airport ao -- Dimensión de aeropuerto de origen
  ON ff.origin_airport_id = ao.airport_id
INNER JOIN airline_catalog.gold.dim_airport ad -- Dimensión de aeropuerto de destino
  ON ff.destination_airport_id = ad.airport_id
INNER JOIN airline_catalog.gold.dim_date dt -- Dimensión de fecha
  ON ff.date_id = dt.date_id
```
