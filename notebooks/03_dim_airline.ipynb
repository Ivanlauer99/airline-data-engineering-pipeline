```sql
-- Carga de datos en la dimensión airline: inserta todas las aerolíneas desde silver y lookup
INSERT INTO airline_catalog.gold.dim_airline (airline_code, airline_name)
SELECT 
    DISTINCT source.airline_code, -- Código de aerolínea
    lookup.Description AS airline_name -- Nombre de la aerolínea obtenido del catálogo de lookups
FROM airline_catalog.silver.flights_silver source
LEFT JOIN airline_catalog.lookups.airline_lookup lookup
ON source.airline_code = lookup.code -- Obtiene el nombre de la aerolínea desde el catálogo de lookups
WHERE source.airline_code IS NOT NULL -- Filtra códigos válidos
  AND lookup.Description IS NOT NULL; -- Solo inserta si existe descripción

-- Consulta para validar las aerolíneas nuevas y sus nombres antes de la inserción
SELECT DISTINCT source.airline_code, lookup.Description AS airline_name
FROM airline_catalog.silver.flights_silver source
LEFT JOIN airline_catalog.lookups.airline_lookup lookup
ON source.airline_code = lookup.code -- Obtiene el nombre de la aerolínea desde el catálogo de lookups
WHERE source.airline_code IS NOT NULL -- Filtra códigos válidos
  AND lookup.Description IS NOT NULL; -- Solo muestra si existe descripción
```
