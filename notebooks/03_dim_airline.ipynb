```sql
--Carga de datos en la dimension airline 
INSERT INTO airline_catalog.gold.dim_airline (airline_code)
SELECT DISTINCT source.airline_code
FROM (
    SELECT airline_code
    FROM airline_catalog.silver.flights_silver
) source
LEFT ANTI JOIN airline_catalog.gold.dim_airline target
ON source.airline_code = target.airline_code
WHERE source.airline_code IS NOT NULL;
```
