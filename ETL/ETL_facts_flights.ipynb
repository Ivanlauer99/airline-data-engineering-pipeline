```sql
--Insertar datos a la fact de gold desde silver 
MERGE INTO airline_catalog.gold.fact_flights t
USING (

    SELECT
        s.flight_id,
        s.flight_date AS date_id,

        a.airline_id,
        ao.airport_id AS origin_airport_id,
        ad.airport_id AS destination_airport_id,

        s.flight_number,
        s.tail_number,

        s.scheduled_departure_time,
        s.actual_departure_time,
        s.departure_delay,

        s.scheduled_arrival_time,
        s.actual_arrival_time,
        s.arrival_delay,

        s.cancelled,
        s.cancellation_code,
        s.diverted,

        s.actual_elapsed_time,
        s.air_time,
        s.distance,

        s.carrier_delay,
        s.weather_delay,
        s.nas_delay,
        s.security_delay,
        s.late_aircraft_delay,

        s.flight_status,

        CURRENT_TIMESTAMP() AS _processing_timestamp

    FROM airline_catalog.silver.flights_silver s

    LEFT JOIN airline_catalog.gold.dim_airline a
        ON s.airline_code = a.airline_code

    LEFT JOIN airline_catalog.gold.dim_airport ao
        ON s.origin_code = ao.airport_code

    LEFT JOIN airline_catalog.gold.dim_airport ad
        ON s.destination_code = ad.airport_code

) src

ON t.flight_id = src.flight_id

WHEN MATCHED THEN UPDATE SET *

WHEN NOT MATCHED THEN INSERT *;
```
