```sql
-- Consulta de desempeño de aerolíneas: total de vuelos, promedio de retraso y porcentaje de vuelos a tiempo
SELECT
    a.airline_name, -- Nombre de la aerolínea
    COUNT(*) AS total_flights, -- Total de vuelos operados por la aerolínea
    ROUND(AVG(f.departure_delay),2) AS avg_delay_minutes, -- Promedio de retraso en salida (minutos)
    ROUND(
        SUM(
            CASE 
                WHEN f.departure_delay <= 0 THEN 1 -- Vuelos sin retraso o con salida anticipada
                ELSE 0 
            END
        ) / COUNT(*) * 100
    ,2) AS on_time_percentage -- Porcentaje de vuelos a tiempo

FROM airline_catalog.gold.fact_flights f

INNER JOIN airline_catalog.gold.dim_airline a
ON f.airline_id = a.airline_id -- Unión por identificador de aerolínea

GROUP BY a.airline_name

ORDER BY avg_delay_minutes DESC; -- Ordenar por mayor promedio de retraso
```
