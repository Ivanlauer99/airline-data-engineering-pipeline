```sql
SELECT
    -- Total de vuelos operados
    COUNT(*) AS total_vuelos,

    -- Promedio de retraso en minutos (solo vuelos no cancelados ni desviados)
    ROUND(
        AVG(CASE WHEN cancelled = false AND diverted = false THEN arrival_delay ELSE NULL END),
        2
    ) AS promedio_retraso_minutos,

    -- Porcentaje de vuelos a tiempo (retraso en llegada <= 0, solo vuelos no cancelados ni desviados)
    ROUND(
        SUM(
            CASE 
                WHEN cancelled = false AND diverted = false AND arrival_delay <= 0 THEN 1
                ELSE 0
            END
        ) / 
        COUNT(*) * 100,
        2
    ) AS porcentaje_vuelos_a_tiempo,

    -- Porcentaje de vuelos cancelados
    ROUND(
        SUM(CASE WHEN cancelled = true THEN 1 ELSE 0 END) / COUNT(*) * 100,
        2
    ) AS porcentaje_vuelos_cancelados,

    -- Porcentaje de vuelos desviados
    ROUND(
        SUM(CASE WHEN diverted = true THEN 1 ELSE 0 END) / COUNT(*) * 100,
        2
    ) AS porcentaje_vuelos_desviados

FROM airline_catalog.gold.fact_flights;
```
