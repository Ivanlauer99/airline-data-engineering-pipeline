```sql
-- Consulta para obtener el nombre del aeropuerto, total de salidas y promedio de minutos de retraso
SELECT
    ap.airport_name, -- Nombre del aeropuerto
    COUNT(*) AS total_salidas, -- Total de salidas desde el aeropuerto
    ROUND(AVG(f.departure_delay), 2) AS promedio_minutos_retraso -- Promedio de minutos de retraso en salida (redondeado a 2 decimales)
FROM airline_catalog.gold.fact_flights f
INNER JOIN airline_catalog.gold.dim_airport ap
    ON f.origin_airport_id = ap.airport_id
WHERE f.departure_delay IS NOT NULL -- Filtrar solo vuelos con información de retraso
GROUP BY ap.airport_name
ORDER BY promedio_minutos_retraso DESC; -- Ordenar por promedio de retraso descendente
```
