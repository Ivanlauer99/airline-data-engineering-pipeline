```sql
--Creacion de la tabla Fact
--Flight_ID (PK), date_id (FK), airline_id (FK), origin_airport_id (FK), destination_airport_id (FK)

CREATE TABLE IF NOT EXISTS airline_catalog.gold.fact_flights
(
    flight_id BIGINT
    COMMENT 'Identificador único del vuelo (surrogate key del dataset)',

    date_id DATE
    COMMENT 'Fecha del vuelo (FK hacia dim_date)',

    airline_id BIGINT
    COMMENT 'FK hacia dim_airline (aerolínea operadora)',

    origin_airport_id BIGINT
    COMMENT 'FK hacia dim_airport (aeropuerto de origen)',

    destination_airport_id BIGINT
    COMMENT 'FK hacia dim_airport (aeropuerto de destino)',

    flight_number INT
    COMMENT 'Número comercial del vuelo',

    tail_number STRING
    COMMENT 'Identificador de la aeronave',

    scheduled_departure_time INT
    COMMENT 'Hora programada de salida (HHMM)',

    actual_departure_time INT
    COMMENT 'Hora real de salida (HHMM)',

    departure_delay DOUBLE
    COMMENT 'Retraso en salida (minutos)',

    scheduled_arrival_time INT
    COMMENT 'Hora programada de llegada (HHMM)',

    actual_arrival_time INT
    COMMENT 'Hora real de llegada (HHMM)',

    arrival_delay DOUBLE
    COMMENT 'Retraso en llegada (minutos)',

    cancelled BOOLEAN
    COMMENT 'Indica si el vuelo fue cancelado',

    cancellation_code STRING
    COMMENT 'Motivo de cancelación (si aplica)',

    diverted BOOLEAN
    COMMENT 'Indica si el vuelo fue desviado',

    actual_elapsed_time DOUBLE
    COMMENT 'Duración total del vuelo en minutos',

    air_time DOUBLE
    COMMENT 'Tiempo real en aire (minutos)',

    distance DOUBLE
    COMMENT 'Distancia recorrida (millas)',

    carrier_delay DOUBLE
    COMMENT 'Retraso causado por la aerolínea',

    weather_delay DOUBLE
    COMMENT 'Retraso causado por clima',

    nas_delay DOUBLE
    COMMENT 'Retraso por sistema aéreo/nacional',

    security_delay DOUBLE
    COMMENT 'Retraso por seguridad',

    late_aircraft_delay DOUBLE
    COMMENT 'Retraso por llegada tardía de aeronave anterior',

    flight_status STRING
    COMMENT 'Estado del vuelo (On Time, Delayed, Cancelled)',

    _processing_timestamp TIMESTAMP
    COMMENT 'Timestamp de carga del registro en el pipeline'
)
USING DELTA
COMMENT 'Fact table de vuelos - capa GOLD (modelo estrella para analítica)'
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported');
```
