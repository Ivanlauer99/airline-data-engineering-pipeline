```python
%sql
-- Crear tabla EXTERNA leyendo el CSV 
 DROP TABLE IF EXISTS airline_catalog.bronze.flights_bronze;
 CREATE TABLE airline_catalog.bronze.flights_bronze
 SELECT * FROM read_files(
 '/Volumes/airline_catalog/landing/archivos/*.csv',
 format => 'csv',
 header => true
 )
 ;
```

```python
%sql
SELECT COUNT(*) AS total_registros FROM airline_catalog.bronze.flights_bronze;
```

```python
%sql
SELECT day_of_month, COUNT(*) AS total_registros FROM airline_catalog.bronze.flights_bronze
GROUP BY DAY_OF_MONTH
ORDER BY DAY_OF_MONTH
```
