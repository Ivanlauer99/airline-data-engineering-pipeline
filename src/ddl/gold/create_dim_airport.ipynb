```sql
-- Creación de tabla dim_airport con Surrogate Key
DROP TABLE IF EXISTS airline_catalog.gold.dim_airport;
CREATE TABLE IF NOT EXISTS airline_catalog.gold.dim_airport
(
    airport_sk BIGINT GENERATED ALWAYS AS IDENTITY 
        COMMENT 'Surrogate key autoincrementado (PK)',
    airport_id BIGINT 
        COMMENT 'ID del aeropuerto (business key del dataset)',
    airport_code STRING 
        COMMENT 'Código IATA del aeropuerto (JFK, LAX, ORD)',
    airport_name STRING 
        COMMENT 'Nombre completo del aeropuerto',
    city STRING 
        COMMENT 'Ciudad donde se ubica el aeropuerto',
    state STRING 
        COMMENT 'Estado donde se ubica el aeropuerto',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
        COMMENT 'Timestamp de creación del registro'
)
USING DELTA
TBLPROPERTIES(
    'delta.feature.allowColumnDefaults' = 'supported',
    'delta.enableChangeDataFeed' = 'true'
)
COMMENT 'Dimensión de aeropuertos - Capa GOLD (modelo estrella)';

-- Nota: airport_sk es el SK que se usa en fact_flights como FK
-- airport_code es la business key usada en el MERGE para identificar registros únicos
```
