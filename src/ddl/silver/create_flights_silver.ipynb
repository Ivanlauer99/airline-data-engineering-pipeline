```sql
--Creacion de la tabla Silver desde la tabla Bronze
CREATE TABLE IF NOT EXISTS airline_catalog.silver.flights_silver
(
    flight_id BIGINT GENERATED ALWAYS AS IDENTITY
        COMMENT 'ID único del vuelo auto-generado',

    flight_date DATE
        COMMENT 'Fecha del vuelo',

    airline_code STRING
        COMMENT 'Código de la aerolínea operadora',

    tail_number STRING
        COMMENT 'Matrícula de la aeronave',

    flight_number INT
        COMMENT 'Número de vuelo',

    origin_airport_id INT
        COMMENT 'Identificador del aeropuerto de origen',

    origin_code STRING
        COMMENT 'Código del aeropuerto de origen',

    origin_city STRING
        COMMENT 'Ciudad del aeropuerto de origen',

    origin_state STRING
        COMMENT 'Estado del aeropuerto de origen',

    destination_airport_id INT
        COMMENT 'Identificador del aeropuerto destino',

    destination_code STRING
        COMMENT 'Código del aeropuerto destino',

    destination_city STRING
        COMMENT 'Ciudad del aeropuerto destino',

    destination_state STRING
        COMMENT 'Estado del aeropuerto destino',

    scheduled_departure_time STRING
        COMMENT 'Hora programada de salida (HH:MM)',

    actual_departure_time STRING
        COMMENT 'Hora real de salida (HH:MM)',

    departure_delay DOUBLE
        COMMENT 'Retraso de salida en minutos',

    scheduled_arrival_time STRING
        COMMENT 'Hora programada de llegada (HH:MM)',

    actual_arrival_time STRING
        COMMENT 'Hora real de llegada (HH:MM)',

    arrival_delay DOUBLE
        COMMENT 'Retraso de llegada en minutos',

    cancelled BOOLEAN
        COMMENT 'Indica si el vuelo fue cancelado',

    cancellation_code STRING
        COMMENT 'Código de motivo de cancelación',

    diverted BOOLEAN
        COMMENT 'Indica si el vuelo fue desviado',

    actual_elapsed_time DOUBLE
        COMMENT 'Tiempo total del vuelo en minutos',

    air_time DOUBLE
        COMMENT 'Tiempo efectivo en aire en minutos',

    distance DOUBLE
        COMMENT 'Distancia recorrida en millas',

    carrier_delay DOUBLE
        COMMENT 'Retraso causado por la aerolínea',

    weather_delay DOUBLE
        COMMENT 'Retraso causado por clima',

    nas_delay DOUBLE
        COMMENT 'Retraso asociado al sistema aeronáutico',

    security_delay DOUBLE
        COMMENT 'Retraso por seguridad',

    late_aircraft_delay DOUBLE
        COMMENT 'Retraso por aeronave anterior',

    route STRING
        COMMENT 'Ruta origen-destino',

    flight_status STRING
        COMMENT 'Estado del vuelo: On Time, Delayed o Cancelled',

    _source_table STRING DEFAULT 'airline_catalog.bronze.flights_bronze'
        COMMENT 'Tabla origen',

    _processing_timestamp TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
        COMMENT 'Timestamp de procesamiento'
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Vuelos - Capa Silver (datos limpios y validados)';
```
